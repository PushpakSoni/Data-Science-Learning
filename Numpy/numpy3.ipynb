```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Data structure: ["restaurant_id", 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 15000, 16000, 17000, 18000],
    [2, 12000, 13500, 14800, 16200],
    [3, 18000, 19200, 20500, 22000],
    [4, 9500, 11000, 12500, 14000],
    [5, 21000, 22500, 24000, 26000]
])

print("===== Zomato Restaurant Sales Data =====")
print('\n Sample data for 1st 3 restauant\n', sales_data[:3])  # Displaying first 3 records)
```

```python
# total sales per year
total_sales_per_year = np.sum(sales_data[:, 1:], axis=0)
print("Total sales per year:", total_sales_per_year)


total_Sales_per_restaurant = np.sum(sales_data[:, 1:], axis=1)
print("Total sales per restaurant:", total_Sales_per_restaurant)
```

```python
# Minimum sales per rest
minimumSales_per_restaurant = np.min(sales_data[:, 1:], axis=1)
print("Minimum sales per restaurant:", minimumSales_per_restaurant)
```

```python
# maximum sales per rest
maxi_sale_per_year = np.max(sales_data[:,1:], axis=0)
print("Maximum sales per year:", maxi_sale_per_year)
```

```python
# avg sale per restaurant
avg_sale_per_restaurant = np.mean(sales_data[:, 1:], axis=1)
print("Average sales per restaurant:", avg_sale_per_restaurant)
```

```python
# cummulative sale per restaurant
cumulative_sales_per_restaurant = np.cumsum(sales_data[:, 1:], axis=1)
print("Cumulative sales per restaurant:\n", cumulative_sales_per_restaurant)

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumulative_sales_per_restaurant, axis=0))
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
